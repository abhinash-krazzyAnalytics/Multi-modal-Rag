```python
from google.colab import files

uploaded = files.upload()

pdf_path = next(iter(uploaded))
print(pdf_path)
```

```python
import fitz

doc = fitz.open(pdf_path)

pages = []

for page_number, page in enumerate(doc):
    text = page.get_text()

    pages.append({
        "page": page_number + 1,
        "text": text
    })

print(f"Pages: {len(pages)}")
print(pages[0]["text"][:1000])
```

```python
import pdfplumber

tables = []

with pdfplumber.open(pdf_path) as pdf:

    for page_number, page in enumerate(pdf.pages, start=1):

        extracted_tables = page.extract_tables()

        for table_number, table in enumerate(extracted_tables, start=1):

            tables.append({
                "page": page_number,
                "table_number": table_number,
                "data": table
            })

print("Tables found:", len(tables))
```

```python
def table_to_text(table):

    rows = []

    for row in table["data"]:
        rows.append(" | ".join(
            str(cell) if cell is not None else ""
            for cell in row
        ))

    return (
        f"Table {table['table_number']} "
        f"from page {table['page']}:\n"
        + "\n".join(rows)
    )
```

```python
table_documents = [
    table_to_text(table)
    for table in tables
]

print(table_documents[0])
```

```python
from langchain_core.documents import Document

documents = []

# Text
for page in pages:

    documents.append(
        Document(
            page_content=page["text"],
            metadata={
                "source": pdf_path,
                "type": "text",
                "page": page["page"]
            }
        )
    )

# Tables
for table in tables:

    documents.append(
        Document(
            page_content=table_to_text(table),
            metadata={
                "source": pdf_path,
                "type": "table",
                "page": table["page"],
                "table": table["table_number"]
            }
        )
    )

print("Documents:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150
)

chunks = splitter.split_documents(documents)

print("Chunks:", len(chunks))
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are an intelligent document analysis assistant.

Answer the user's question using ONLY the information provided
in the retrieved document context.

Rules:
1. Do not invent information.
2. If the answer is not present, say:
   "I could not find this information in the document."
3. Pay special attention to tables and numerical values.
4. Preserve numbers, dates, percentages and units accurately.
5. Mention the page number when available.
6. If multiple sources provide conflicting information, explain the conflict.

Retrieved Context:
{context}

User Question:
{question}

Answer:
""")
```

```python
def ask_rag(question, k=5):

    # Retrieve relevant chunks
    docs = vector_store.similarity_search(
        question,
        k=k
    )

    # Build context
    context_parts = []

    for doc in docs:

        source_type = doc.metadata.get("type", "unknown")
        page = doc.metadata.get("page", "unknown")

        context_parts.append(
            f"""
[Source Type: {source_type}]
[Page: {page}]

{doc.page_content}
"""
        )

    context = "\n\n".join(context_parts)

    # Create prompt
    messages = prompt.format_messages(
        context=context,
        question=question
    )

    # Ask Ollama
    response = llm.invoke(messages)

    return response.content, docs
```

```python
answer, sources = ask_rag(
    "ApexNova Refining & Energy Ltd founded?"
)

print(answer)
```
